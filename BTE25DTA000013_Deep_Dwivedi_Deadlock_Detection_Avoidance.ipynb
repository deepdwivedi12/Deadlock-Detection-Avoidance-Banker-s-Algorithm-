## 1. Sample Input

We use 5 processes and 3 resource types: A, B and C.

- Allocation: resources currently given to each process.
- Max: maximum resources each process may need.
- Available: resources currently free.

```python
# Allocation Matrix
allocation = [
    [0, 1, 0],
    [2, 0, 0],
    [3, 0, 2],
    [2, 1, 1],
    [0, 0, 2]
]

# Maximum Matrix
maximum = [
    [7, 5, 3],
    [3, 2, 2],
    [9, 0, 2],
    [2, 2, 2],
    [4, 3, 3]
]

# Available Resources
available = [3, 3, 2]

print("Allocation Matrix:")
for row in allocation:
    print(row)

print("\nMaximum Matrix:")
for row in maximum:
    print(row)

print("\nAvailable Resources:", available)
```

## 2. Calculate Need Matrix
Need = Maximum - Allocation

```python
need = []

for i in range(5):
    row = []
    for j in range(3):
        row.append(maximum[i][j] - allocation[i][j])
    need.append(row)

print("Need Matrix:")
for row in need:
    print(row)
```

## 3. Banker's Algorithm – Safe State

The algorithm checks whether each process can get its remaining resources and finish.

If a process finishes, it releases its allocated resources. We add those resources to Available.

```python
work = available.copy()
finish = [False, False, False, False, False]
safe_sequence = []

while len(safe_sequence) < 5:
    found = False

    for i in range(5):
        if finish[i] == False:
            if (need[i][0] <= work[0] and
                need[i][1] <= work[1] and
                need[i][2] <= work[2]):

                # Process finishes and releases its resources
                work[0] += allocation[i][0]
                work[1] += allocation[i][1]
                work[2] += allocation[i][2]

                finish[i] = True
                safe_sequence.append(i)
                found = True

    if found == False:
        break

if len(safe_sequence) == 5:
    print("System is in SAFE STATE")
    print("Safe Sequence:", end=" ")
    for p in safe_sequence:
        print("P" + str(p), end=" ")
else:
    print("System is in UNSAFE STATE")
```

## 4. Deadlock Detection – Simple Example

For detection, we use:
- Allocation: resources currently held by processes.
- Request: resources each process is still waiting for.
- Available: free resources.

A process can finish if its Request is less than or equal to Available. When it finishes, it releases its allocated resources.

```python
# Two processes and two resource types
allocation2 = [
    [1, 0],
    [0, 1]
]

request = [
    [0, 1],
    [1, 0]
]

available2 = [0, 0]

finish2 = [False, False]
work2 = available2.copy()

for i in range(2):
    if allocation2[i] == [0, 0]:
        finish2[i] = True

for i in range(2):
    if finish2[i] == False:
        if (request[i][0] <= work2[0] and
            request[i][1] <= work2[1]):

            work2[0] += allocation2[i][0]
            work2[1] += allocation2[i][1]
            finish2[i] = True

if finish2[0] == False or finish2[1] == False:
    print("Deadlock Detected")
else:
    print("No Deadlock Detected")
```
